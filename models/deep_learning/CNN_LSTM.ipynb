## Validation Metrics Summary

```python
import pandas as pd
from src import config

problems = ['S1', 'S24', 'D1', 'D24']
rows = []
for pid in problems:
    r = eval(f"results_{pid}")
    row = {"model": "cnn_lstm", "problem": pid}
    row.update(r["metrics"])
    rows.append(row)
    # Save predictions parquet
    pred_df = pd.DataFrame({
        "timestamp": r["ts"],
        "y_true_mw": r["y_true"],
        "y_pred_mw": r["y_pred"],
    }).set_index("timestamp")
    pred_path = config.RESULTS_PREDICTIONS_DIR / f"cnn_lstm_{pid}_val_predictions.parquet"
    pred_df.to_parquet(pred_path)
    print(f"Saved predictions -> {pred_path.name}")

summary_df = pd.DataFrame(rows)
csv_path = config.RESULTS_TABLES_DIR / f"cnn_lstm_val_metrics.csv"
summary_df.to_csv(csv_path, index=False)
print("\nMetrics CSV saved:", csv_path.name)
print(summary_df.to_string(index=False))
```
